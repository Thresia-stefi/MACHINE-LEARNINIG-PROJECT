```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer

categories = [
    'sci.space',
    'comp.graphics',
    'rec.sport.baseball'
]

dataset = fetch_20newsgroups(
    subset='train',
    categories=categories,
    remove=('headers', 'footers', 'quotes')
)
print("Number of Documents:", len(dataset.data))

vectorizer = CountVectorizer(
    stop_words='english',
    max_features=1000
)
X = vectorizer.fit_transform(dataset.data)
print(X.shape)

word_counts = np.array(X.sum(axis=0)).flatten()
vocabulary = vectorizer.get_feature_names_out()
print("Vocabulary Size:", len(vocabulary))

total_words = np.sum(word_counts)
mle_probs = word_counts / total_words


alpha = 0.1
map_probs_01 = (
    word_counts + alpha
) / (
    total_words + alpha * len(word_counts)
)

alpha = 1
map_probs_1 = (
    word_counts + alpha
) / (
    total_words + alpha * len(word_counts)
)

alpha = 10
map_probs_10 = (
    word_counts + alpha
) / (
    total_words + alpha * len(word_counts)
)

comparison = pd.DataFrame({
    "Word": vocabulary[:20],
    "MLE": mle_probs[:20],
    "MAP α=0.1": map_probs_01[:20],
    "MAP α=1": map_probs_1[:20],
    "MAP α=10": map_probs_10[:20]
})
print(comparison)


plt.figure(figsize=(12, 6))
plt.plot(mle_probs[:20], label="MLE")
plt.plot(map_probs_01[:20], label="MAP α=0.1")
plt.plot(map_probs_1[:20], label="MAP α=1")
plt.plot(map_probs_10[:20], label="MAP α=10")
plt.legend()
plt.title("MLE vs MAP Probability Estimates")
plt.xlabel("Words")
plt.ylabel("Probability")
plt.show()


top = np.argsort(word_counts)[::-1][:10]
top_df = pd.DataFrame({
    "Word": vocabulary[top],
    "Count": word_counts[top],
    "MLE Probability": mle_probs[top]
})
print(top_df)
```
